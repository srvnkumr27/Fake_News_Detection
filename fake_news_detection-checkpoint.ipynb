```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import string
import ast

from scipy.sparse import hstack, csr_matrix

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.preprocessing import StandardScaler

from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

sns.set_style("whitegrid")
RANDOM_STATE = 42
```

```python
df = pd.read_csv("train.csv")
print("Shape:", df.shape)
df.head()
```

```python
df.fillna("", inplace=True)

# combine title + body into one text field
df["content_raw"] = df["title"].astype(str) + " " + df["full_text"].astype(str)
df[["title", "content_raw"]].head()
```

```python
# A compact, explicit manual stopword list (not delegated to the vectorizer)
_stopword_text = (
    "a an the and or but if while is are was were be been being "
    "to of in on at by for with about against between into through "
    "during before after above below from up down out off over under "
    "again further then once here there when where why how all any "
    "both each few more most other some such no nor not only own same "
    "so than too very s t can will just don should now this that "
    "these those i you he she it we they them his her its our their "
    "as do does did doing have has had having"
)
MANUAL_STOPWORDS = set(_stopword_text.split())

def clean_text(text: str) -> str:
    """Lowercase, strip URLs/HTML/punctuation/digits, collapse whitespace."""
    text = text.lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)      # URLs
    text = re.sub(r"<.*?>", " ", text)                  # HTML tags
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = re.sub(r"\d+", " ", text)                    # digits
    text = re.sub(r"\s+", " ", text).strip()
    return text

def manual_tokenize(text: str) -> list:
    """Simple whitespace tokenizer."""
    return text.split()

def remove_stopwords(tokens: list) -> list:
    """Manual stopword removal against our own stopword set."""
    return [tok for tok in tokens if tok not in MANUAL_STOPWORDS and len(tok) > 1]

def preprocess(text: str) -> str:
    cleaned = clean_text(text)
    tokens = manual_tokenize(cleaned)
    tokens = remove_stopwords(tokens)
    return " ".join(tokens)

df["content"] = df["content_raw"].apply(preprocess)
df["token_count"] = df["content"].apply(lambda x: len(x.split()))
df[["content_raw", "content", "token_count"]].head()
```

```python
from collections import Counter
def top_words(texts,n=15):
    counter = Counter()
    for t in texts:
        counter.update(t.split())
    return counter.most_common(n)
real_top = top_words(df.loc[df.label == 0, "content"])
fake_top = top_words(df.loc[df.label == 1, "content"])
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.barplot(x=[c for _, c in real_top], y=[w for w, _ in real_top], ax=axes[0], color="steelblue")
axes[0].set_title("Top Words — Real (label 0)")
sns.barplot(x=[c for _, c in fake_top], y=[w for w, _ in fake_top], ax=axes[1], color="indianred")
axes[1].set_title("Top Words — Fake (label 1)")
plt.tight_layout()
plt.show()
```

```python
bow_vectorizer = CountVectorizer(max_features=5000)
X_bow = bow_vectorizer.fit_transform(df["content"])

# 2) TF-IDF
tfidf_vectorizer = TfidfVectorizer(max_features=5000)
X_tfidf = tfidf_vectorizer.fit_transform(df["content"])

print("Bag-of-Words shape:", X_bow.shape)
print("TF-IDF shape:      ", X_tfidf.shape)
```

```python
def parse_embedding(cell):
    if isinstance(cell, str):
        return np.array(ast.literal_eval(cell), dtype=float)
    return np.array(cell, dtype=float)

embedding_matrix = np.vstack(df["text_embedding"].apply(parse_embedding).values)
print("Embedding matrix shape:", embedding_matrix.shape)
```

```python
scaler = StandardScaler()
numeric_scaled = scaler.fit_transform(df[numeric_columns].astype(float))

y = df["label"].values
print("numeric_scaled shape:", numeric_scaled.shape)
```

```python
def make_models():
    return {
        "KNN": KNeighborsClassifier(n_neighbors=5),
        "Logistic Regression": LogisticRegression(max_iter=1000),
        "Random Forest": RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE),
        "Neural Network": MLPClassifier(hidden_layer_sizes=(100,), max_iter=300, random_state=RANDOM_STATE),
    }

def evaluate(models, X_train, X_test, y_train, y_test, label=""):
    rows = []
    preds_by_model = {}
    for name, model in models.items():
        model.fit(X_train, y_train)
        preds = model.predict(X_test)
        preds_by_model[name] = preds
        acc = accuracy_score(y_test, preds)
        rows.append({"Feature Set": label, "Model": name, "Accuracy": acc})
        print(f"[{label}] {name} — Accuracy: {acc:.4f}")
        print(classification_report(y_test, preds))
    return pd.DataFrame(rows), preds_by_model
```

```python
Xa_train, Xa_test, ya_train, ya_test = train_test_split(
    X_tfidf, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
results_a, preds_a = evaluate(make_models(), Xa_train, Xa_test, ya_train, ya_test, label="Text (TF-IDF) only")
```

```python
Xb_train, Xb_test, yb_train, yb_test = train_test_split(
    numeric_scaled, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
results_b, preds_b = evaluate(make_models(), Xb_train, Xb_test, yb_train, yb_test, label="Numeric scores only")
```

```python
X_full = hstack([X_tfidf, numeric_scaled])
Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X_full, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
results_c, preds_c = evaluate(make_models(), Xc_train, Xc_test, yc_train, yc_test, label="Full (TF-IDF + numeric)")
```

```python
fig, axes = plt.subplots(1, 4, figsize=(20, 4))
for ax, (name, preds) in zip(axes, preds_c.items()):
    ConfusionMatrixDisplay.from_predictions(yc_test, preds, ax=ax, colorbar=False)
    ax.set_title(name)
plt.tight_layout()
plt.show()
```

```python
all_results = pd.concat([results_a, results_b, results_c], ignore_index=True)
pivot = all_results.pivot(index="Model", columns="Feature Set", values="Accuracy")
pivot
```

```python
plt.figure(figsize=(9, 5))
sns.barplot(data=all_results, x="Model", y="Accuracy", hue="Feature Set")
plt.title("Model Accuracy by Feature Set")
plt.ylim(0, 1.05)
plt.xticks(rotation=15)
plt.legend(title="Feature Set", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
rf_full = RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE)
rf_full.fit(Xc_train, yc_train)

feature_names = list(tfidf_vectorizer.get_feature_names_out()) + numeric_columns
importances = rf_full.feature_importances_

top_idx = np.argsort(importances)[::-1][:15]
top_features = pd.DataFrame({
    "feature": [feature_names[i] for i in top_idx],
    "importance": importances[top_idx],
})
top_features
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(data=top_features, x="importance", y="feature", color="teal")
plt.title("Top 15 Feature Importances — Random Forest (Full Feature Set)")
plt.tight_layout()
plt.show()
```

```python
def predict_sample(raw_text: str, numeric_feature_values: dict, model, tfidf=tfidf_vectorizer,
                    scaler=scaler, numeric_columns=numeric_columns):
    cleaned = preprocess(raw_text)
    text_vec = tfidf.transform([cleaned])
    numeric_vec = scaler.transform(pd.DataFrame([numeric_feature_values])[numeric_columns])
    features = hstack([text_vec, numeric_vec])
    return model.predict(features)[0]

sample_text = "Government announces a new policy for higher education."
sample_numeric = {
    "sentiment_score": 0.1,
    "clickbait_score": 0.15,
    "emotional_word_ratio": 0.12,
    "readability_score": 65,
    "source_trust_score": 0.9,
}

pred = predict_sample(sample_text, sample_numeric, rf_full)
print("Predicted label:", pred, "(0 = Real, 1 = Fake)")
```
